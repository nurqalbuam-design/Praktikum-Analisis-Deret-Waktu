```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

import warnings
warnings.filterwarnings("ignore")
```

# MA(1)


## No 1
Buatlah simulasi deret MA (1) dengan panjang pengamatan 20 dengan nilai parameter θ=0.1,  θ=0.3, θ=0.7, θ=1.0, θ=1.5, θ=2.0

```python
np.random.seed(170)
ar = np.array([1])
# θ = 0.1
ma_01 = np.array([1, -0.1])
model_ma_01 = ArmaProcess(ar, ma_01)
data_ma_01 = model_ma_01.generate_sample(nsample=20)

# θ = 0.3
ma_03 = np.array([1, -0.3])
model_ma_03 = ArmaProcess(ar, ma_03)
data_ma_03 = model_ma_03.generate_sample(nsample=20)

# θ = 0.7
ma_07 = np.array([1, -0.7])
model_ma_07 = ArmaProcess(ar, ma_07)
data_ma_07 = model_ma_07.generate_sample(nsample=20)

# θ = 1.0
ma_10 = np.array([1, -1.0])
model_ma_10 = ArmaProcess(ar, ma_10)
data_ma_10 = model_ma_10.generate_sample(nsample=20)

# θ = 1.5
ma_15 = np.array([1, -1.5])
model_ma_15 = ArmaProcess(ar, ma_15)
data_ma_15 = model_ma_15.generate_sample(nsample=20)

# θ = 2.0
ma_20 = np.array([1, -2.0])
model_ma_20 = ArmaProcess(ar, ma_20)
data_ma_20 = model_ma_20.generate_sample(nsample=20)
```

```python
hasil_ma1 = pd.DataFrame({
    "theta_0.1": data_ma_01,
    "theta_0.3": data_ma_03,
    "theta_0.7": data_ma_07,
    "theta_1.0": data_ma_10,
    "theta_1.5": data_ma_15,
    "theta_2.0": data_ma_20
})

hasil_ma1
```

## No 2
Buatlah plot ACF dari masing-masing deret hasil simulasi yang Anda peroleh pada nomor 1 dan bandingkan. Berdasarkan plot ACF tersebut, apa yang dapat Anda simpulkan terkait nilai autokorelasi pada lag 1 pada model MA(1)? (Hint: Identifikasi pola nilai autokorelasi berdasarkan nilai parameter MA, kaitkan nilai autokorelasi sampel dengan nilai autokorelasi MA(1) secara teori.)



```python
fig, axes = plt.subplots(3, 2, figsize=(12, 10))

plot_acf(data_ma_01, lags=10, ax=axes[0,0])
axes[0,0].set_title("ACF MA(1), θ = 0.1")

plot_acf(data_ma_03, lags=10, ax=axes[0,1])
axes[0,1].set_title("ACF MA(1), θ = 0.3")

plot_acf(data_ma_07, lags=10, ax=axes[1,0])
axes[1,0].set_title("ACF MA(1), θ = 0.7")

plot_acf(data_ma_10, lags=10, ax=axes[1,1])
axes[1,1].set_title("ACF MA(1), θ = 1.0")

plot_acf(data_ma_15, lags=10, ax=axes[2,0])
axes[2,0].set_title("ACF MA(1), θ = 1.5")

plot_acf(data_ma_20, lags=10, ax=axes[2,1])
axes[2,1].set_title("ACF MA(1), θ = 2.0")

plt.tight_layout()
plt.show()
```

Interpretasi Plot ACF Model MA(1)

Berdasarkan plot ACF hasil simulasi, nilai autokorelasi sampel pada lag 1 untuk seluruh nilai parameter $\theta$ cenderung bernilai negatif. Hal ini sesuai dengan bentuk model MA(1) yang digunakan, yaitu:

$$
Y_t = e_t - \theta e_{t-1}
$$

Secara teoritis, nilai autokorelasi pada lag 1 untuk model MA(1) adalah:

$$
\rho_1 = \frac{-\theta}{1+\theta^2}
$$

Sedangkan untuk lag yang lebih besar dari 1 berlaku:

$$
\rho_k = 0, \quad k > 1
$$

Berdasarkan rumus tersebut, nilai autokorelasi teoritis pada lag 1 untuk masing-masing parameter adalah:

| Parameter $\theta$ | $\rho_1$ teoritis |
|---|---|
| 0.1 | -0.099 |
| 0.3 | -0.275 |
| 0.7 | -0.470 |
| 1.0 | -0.500 |
| 1.5 | -0.462 |
| 2.0 | -0.400 |

Berdasarkan nilai tersebut, ketika $\theta$ meningkat dari 0.1 hingga 1.0, nilai autokorelasi pada lag 1 menjadi semakin negatif atau kekuatan autokorelasinya semakin besar. Nilai autokorelasi teoritis paling negatif terjadi pada $\theta=1$, yaitu sebesar $-0.5$. Namun, ketika $\theta>1$, yaitu pada $\theta=1.5$ dan $\theta=2.0$, nilai autokorelasi pada lag 1 kembali mendekati nol.

Nilai autokorelasi sampel yang diperoleh dari plot ACF tidak selalu sama persis dengan nilai autokorelasi teoritis. Hal ini disebabkan simulasi hanya menggunakan 20 pengamatan, sehingga masih terdapat variasi sampling. Oleh karena itu, nilai autokorelasi sampel dapat lebih besar atau lebih kecil dibandingkan nilai teoritisnya.

Secara teoritis, model MA(1) memiliki pola ACF yang *cut off* setelah lag 1 karena autokorelasi pada lag yang lebih besar dari 1 bernilai nol. Pada hasil simulasi masih terlihat beberapa nilai autokorelasi pada lag setelah lag 1. Hal tersebut dapat terjadi akibat variasi sampel karena jumlah pengamatan yang digunakan relatif sedikit.

**Kesimpulan:** Nilai parameter $\theta$ memengaruhi besar autokorelasi pada lag 1. Untuk $0 < \theta \leq 1$, peningkatan nilai $\theta$ menyebabkan autokorelasi lag 1 semakin negatif. Setelah $\theta>1$, nilai autokorelasi lag 1 kembali mendekati nol. Nilai autokorelasi sampel tidak selalu sama persis dengan nilai teoritis karena simulasi hanya menggunakan 20 pengamatan.

```python
# θ=0.25
# untuk n = 60, 100, 150, dan 300

theta_fixed = 0.25
ma_025 = np.array([1, -theta_fixed])

# n = 60
np.random.seed(170)
model_ma_60 = ArmaProcess(ar, ma_025)
data_ma_60 = model_ma_60.generate_sample(nsample=60)

# n = 100
np.random.seed(170)
model_ma_100 = ArmaProcess(ar, ma_025)
data_ma_100 = model_ma_100.generate_sample(nsample=100)

# n = 150
np.random.seed(170)
model_ma_150 = ArmaProcess(ar, ma_025)
data_ma_150 = model_ma_150.generate_sample(nsample=150)

# n = 300
np.random.seed(170)
model_ma_300 = ArmaProcess(ar, ma_025)
data_ma_300 = model_ma_300.generate_sample(nsample=300)
```

Interpretasi Plot ACF Berdasarkan Panjang Pengamatan

Berdasarkan plot ACF, seluruh deret menggunakan parameter MA(1) yang sama, yaitu $\theta=0.25$, tetapi memiliki panjang pengamatan yang berbeda, yaitu $n=60$, $n=100$, $n=150$, dan $n=300$.

Secara teoritis, autokorelasi pada lag 1 untuk model MA(1) yang digunakan adalah:

$$
\rho_1 = \frac{-\theta}{1+\theta^2}
$$

Dengan $\theta=0.25$, diperoleh:

$$
\rho_1 = \frac{-0.25}{1+(0.25)^2} \approx -0.235
$$

Dengan demikian, nilai autokorelasi teoritis pada lag 1 tetap sama untuk seluruh panjang pengamatan karena parameter $\theta$ yang digunakan tidak berubah.

Pada plot ACF hasil simulasi, nilai autokorelasi sampel pada lag 1 terlihat berbeda untuk setiap panjang pengamatan. Perbedaan tersebut terjadi karena nilai yang ditampilkan pada plot merupakan autokorelasi sampel, sehingga masih dipengaruhi oleh variasi sampling.

Selain itu, terlihat bahwa interval kepercayaan pada plot ACF semakin sempit ketika jumlah pengamatan bertambah. Plot dengan $n=60$ memiliki interval kepercayaan yang relatif lebih lebar, sedangkan pada $n=300$ interval kepercayaannya terlihat paling sempit. Hal ini menunjukkan bahwa semakin besar jumlah pengamatan, ketidakpastian dalam mengestimasi nilai autokorelasi semakin kecil sehingga estimasi autokorelasi menjadi lebih presisi.

Secara teoritis, model MA(1) memiliki autokorelasi nol setelah lag 1. Pada hasil simulasi masih terlihat nilai autokorelasi pada lag-lag setelah lag 1, tetapi sebagian besar berada di dalam interval kepercayaan.

**Kesimpulan:** Perbedaan panjang pengamatan tidak mengubah nilai autokorelasi teoritis pada lag 1 karena nilai $\theta$ tetap sama. Namun, semakin besar jumlah pengamatan, interval kepercayaan ACF semakin sempit sehingga estimasi autokorelasi sampel menjadi lebih presisi.

**CATATAN:** Plot ACF pada keempat simulasi menunjukkan tidak adanya autokorelasi signifikan setelah lag 2, konsisten dengan sifat teoritis proses MA(2) yang ACF-nya cut off setelah lag q=2. Pada kasus c dan d, baik lag 1 maupun lag 2 signifikan, sedangkan pada kasus a dan b hanya lag 2 yang signifikan (lag 1 tidak signifikan) — perbedaan ini wajar terjadi akibat variabilitas sampling pada ukuran sampel n=50, namun tidak mengubah kesimpulan utama bahwa pola cut off pada lag 2 tetap konsisten dengan orde MA(2) yang digunakan dalam simulasi.

Kesimpulan

1. **Pola cut off**: Pada keempat kombinasi parameter, autokorelasi yang signifikan (keluar dari selang kepercayaan) hanya teramati pada lag 1 dan/atau lag 2 — tidak pernah melewati lag 2. Mulai lag 3 dan seterusnya, autokorelasi berada dalam selang kepercayaan sehingga dianggap tidak signifikan (mendekati nol). Pola ini sesuai dengan sifat teoritis proses MA(q), yaitu ACF akan terputus (cut off) setelah lag ke-q — untuk MA(2), signifikansi hanya bertahan sampai maksimal lag 2.

2. **Keterkaitan tanda dan besar autokorelasi dengan nilai parameter**: Besar dan arah (tanda) autokorelasi pada lag 1 dan 2 dipengaruhi oleh besar dan tanda parameter θ₁ dan θ₂. Pada seluruh kasus, autokorelasi lag 2 konsisten bernilai besar dan signifikan (magnitude ~0.40-0.52), mengikuti besar |θ₂| = 0.8 yang lebih besar dibanding |θ₁| = 0.3. Sementara autokorelasi lag 1 hanya signifikan pada kasus c dan d (θ₂ = -0.8), dan tidak signifikan pada kasus a dan b (θ₂ = +0.8) — menunjukkan bahwa tanda θ₂ turut memengaruhi seberapa kuat autokorelasi lag 1 muncul secara sampel, sejalan dengan rumus teoritis $\rho_1 = \dfrac{\theta_1(1+\theta_2)}{1+\theta_1^2+\theta_2^2}$ yang dipengaruhi faktor $(1+\theta_2)$.

3. **Implikasi praktis**: Pola cut off pada lag 2 ini merupakan ciri khas yang dapat digunakan untuk mengidentifikasi orde MA (q=2) dari suatu deret waktu — jika plot ACF suatu data menunjukkan signifikansi hanya sampai lag tertentu lalu terputus, itu mengindikasikan proses tersebut adalah MA dengan orde sesuai lag terakhir yang signifikan.

## SOAL NO 10

Simulasi deret AR (2) dengan panjang pengamatan 100 dengan parameter:

a.  $ϕ_1$=1.5,$ϕ_2$=−0.5

b.  $ϕ_1$=0.75,$ϕ_2$=−0.25

c.  $ϕ_1$=1,$ϕ_2$=−0.25

d.  $ϕ_1$=0.5,$ϕ_2$=−0.75

```python
import numpy as np

# =========================================================
# a. phi1 = 1.5, phi2 = -0.5
# =========================================================
np.random.seed(170)

psi1_a, psi2_a = 1.5, -0.5
ar2_a = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(98):
    ar2_a.append(
        psi1_a * ar2_a[i+1]
        + psi2_a * ar2_a[i]
        + np.random.normal(0,1)
    )


# =========================================================
# b. phi1 = 0.75, phi2 = -0.25
# =========================================================
np.random.seed(170)

psi1_b, psi2_b = 0.75, -0.25
ar2_b = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(98):
    ar2_b.append(
        psi1_b * ar2_b[i+1]
        + psi2_b * ar2_b[i]
        + np.random.normal(0,1)
    )


# =========================================================
# c. phi1 = 1, phi2 = -0.25
# =========================================================
np.random.seed(170)

psi1_c, psi2_c = 1, -0.25
ar2_c = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(98):
    ar2_c.append(
        psi1_c * ar2_c[i+1]
        + psi2_c * ar2_c[i]
        + np.random.normal(0,1)
    )


# =========================================================
# d. phi1 = 0.5, phi2 = -0.75
# =========================================================
np.random.seed(170)

psi1_d, psi2_d = 0.5, -0.75
ar2_d = [
    np.random.uniform(0,1),
    np.random.uniform(0,1)
]

for i in range(98):
    ar2_d.append(
        psi1_d * ar2_d[i+1]
        + psi2_d * ar2_d[i]
        + np.random.normal(0,1)
    )
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# a. phi1 = 1.5, phi2 = -0.5
axes[0, 0].plot(ar2_a, '-o')
axes[0, 0].set_xlabel('Waktu')
axes[0, 0].set_ylabel(r'$Y_t$')
axes[0, 0].set_title(r'AR(2): $\phi_1=1.5,\ \phi_2=-0.5$')

# b. phi1 = 0.75, phi2 = -0.25
axes[0, 1].plot(ar2_b, '-o')
axes[0, 1].set_xlabel('Waktu')
axes[0, 1].set_ylabel(r'$Y_t$')
axes[0, 1].set_title(r'AR(2): $\phi_1=0.75,\ \phi_2=-0.25$')

# c. phi1 = 1, phi2 = -0.25
axes[1, 0].plot(ar2_c, '-o')
axes[1, 0].set_xlabel('Waktu')
axes[1, 0].set_ylabel(r'$Y_t$')
axes[1, 0].set_title(r'AR(2): $\phi_1=1,\ \phi_2=-0.25$')

# d. phi1 = 0.5, phi2 = -0.75
axes[1, 1].plot(ar2_d, '-o')
axes[1, 1].set_xlabel('Waktu')
axes[1, 1].set_ylabel(r'$Y_t$')
axes[1, 1].set_title(r'AR(2): $\phi_1=0.5,\ \phi_2=-0.75$')

plt.tight_layout()
plt.show()
```
